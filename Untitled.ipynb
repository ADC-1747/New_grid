## 5. Methodology, Assumptions & AI Disclosure

### Data cleaning approach

1. **Timestamps** — parse to datetime, drop unparseable rows, sort chronologically.
2. **Duplicate rows** — remove exact duplicates, then keep the last record per duplicate timestamp.
3. **Numeric conversion** — coerce all MW, price, and limit columns; non-numeric text becomes missing.
4. **Long format** — reshape Quantity-MW1/Price1 … Quantity-MW13/Price13 into one row per breakpoint.
5. **Missing breakpoints** — drop individual MW/price pairs where either value is missing (keep the hourly row).
6. **Price outliers** — remove points outside **-100 to 500 $/MWh** (typical valid offers for this unit are ~$20–$45/MWh).
7. **MW outliers** — remove negative MW and MW above **250** (unit high limit is ~190–200 MW).
8. **Non-monotonic MW** — sort valid breakpoint pairs by MW within each hour so curves never step backward on the x-axis.
9. **High/Low limits** — swap when Low > High; drop rows that remain invalid after correction.

### Forecast method (target: 05/29/2026)

For each hour of the target day:

1. Use the **same calendar date one year prior** (05/29/2025) when available — used for all 24 hours in this dataset.
2. Else use the **median MW/price at each breakpoint** for the same hour, month, and weekday.
3. Else use the **median MW/price at each breakpoint** for the same hour across all history.

### Assumptions

- Historical offer curves for the same hour-of-day are comparable across days.
- Prior-year same-date is a reasonable baseline when available.
- Fixed outlier thresholds are defensible given the unit's typical operating range.
- `Resource Status` is not used in cleaning or forecasting.
- Duplicate timestamps are resolved by keeping the **last** record after sorting.
